## Cell 3 — LONGER words at native resolution (8-12 letters)

## Read it
- **Short words readable** at native res -> coarseness was the cause. We
  re-extract the full landmark cache at high `T_native` and proceed to the VLM.
- **Still unreadable** even at full res -> the static-image representation is
  fundamentally limited for in-air writing (letters pile; temporal order, not
  static shape, carries the word). Pivot to the sequence model + word-LM.
